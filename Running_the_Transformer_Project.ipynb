```python
import torch
from transformers import pipeline
from sentence_transformers import SentenceTransformer

device = 0 if torch.cuda.is_available() else -1
summarizer = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6", device=device)
qa_model = pipeline("question-answering", model="deepset/roberta-base-squad2", device=device)
embedder = SentenceTransformer("all-MiniLM-L6-v2")
print("All models loaded")
```

```python
from sentence_transformers import util
from pypdf import PdfReader
import gradio as gr

def read_pdf(path):
    reader = PdfReader(path)
    return " ".join((page.extract_text() or "") for page in reader.pages)

def summarize_long(text, chunk_words=400):
    words = text.split()
    chunks = [" ".join(words[i:i + chunk_words]) for i in range(0, len(words), chunk_words)]
    parts = []
    for c in chunks:
        if len(c.split()) < 30:
            continue
        out = summarizer(c, max_length=120, min_length=30, do_sample=False, truncation=True)
        parts.append(out[0]["summary_text"])
    return " ".join(parts)

print("Ready")
```

```python
text = """
Artificial intelligence is transforming industries across the world. Machine learning
models can now read documents, answer questions, translate languages and generate
images. Transformers, introduced in 2017, use self-attention to process whole
sequences in parallel, which made training much faster and more accurate than RNNs.
Today, transformer models power search engines, chatbots, and recommendation systems.
"""

result = summarizer(text, max_length=60, min_length=20, do_sample=False)
print("Summary:", result[0]["summary_text"])
```

```python
context = """
Rahul is a final-year engineering student. He has built projects in Python,
TensorFlow and NLP. He completed a machine learning internship at a startup in 2025.
"""

for q in ["What skills does Rahul have?",
          "What did Rahul complete in 2025?",
          "What year is Rahul studying in?"]:
    ans = qa_model(question=q, context=context)
    print(f"Q: {q}\nA: {ans['answer']}  (confidence: {ans['score']:.2f})\n")
```

```python
from google.colab import files
uploaded = files.upload()   # select one or more resume PDFs

resumes = {name: read_pdf(name) for name in uploaded.keys()}
print("Resumes loaded:", list(resumes.keys()))
```

```python
resumes = {
    "Resume_A": "Python developer with 2 years experience in machine learning, NLP, TensorFlow and data analysis. Built text classification and chatbot projects.",
    "Resume_B": "Java developer skilled in Spring Boot, SQL, REST APIs and web development. Worked on banking applications.",
    "Resume_C": "Data science student with Python, PyTorch, deep learning and NLP projects. Internship in machine learning.",
}
```

```python
job_description = """
We are looking for a Python developer with experience in machine learning,
NLP, deep learning, TensorFlow or PyTorch, and data analysis.
"""

jd_embedding = embedder.encode(job_description, convert_to_tensor=True)

scores = []
for name, text in resumes.items():
    emb = embedder.encode(text[:5000], convert_to_tensor=True)
    score = util.cos_sim(jd_embedding, emb).item()
    scores.append((name, score))

scores.sort(key=lambda x: x[1], reverse=True)

print("Resume Ranking")
print("-" * 40)
for rank, (name, score) in enumerate(scores, 1):
    print(f"{rank}. {name}  ->  match score: {score * 100:.1f}%")
```

```python
questions = {
    "Name": "What is the candidate's name?",
    "Skills": "What technical skills does the candidate have?",
    "Experience": "What work experience does the candidate have?",
    "Education": "What is the candidate's education?",
}

for name, text in resumes.items():
    print(f"\n===== {name} =====")
    for label, q in questions.items():
        ans = qa_model(question=q, context=text[:5000])
        print(f"{label}: {ans['answer']}")
    print("Summary:", summarize_long(text)[:500])
```

```python
def summarize_document(file):
    if file is None:
        return "Please upload a PDF."
    path = file if isinstance(file, str) else file.name
    text = read_pdf(path)
    if len(text.split()) < 30:
        return "Not enough text found in the PDF."
    return summarize_long(text)

def ask_document(file, question):
    if file is None or not question:
        return "Upload a PDF and type a question."
    path = file if isinstance(file, str) else file.name
    text = read_pdf(path)[:8000]
    ans = qa_model(question=question, context=text)
    return f"{ans['answer']}  (confidence: {ans['score']:.2f})"

with gr.Blocks(title="AI Document Summarizer") as app:
    gr.Markdown("## AI Document Summarizer and Q&A")
    pdf = gr.File(label="Upload PDF", file_types=[".pdf"])
    with gr.Tab("Summarize"):
        btn1 = gr.Button("Summarize")
        out1 = gr.Textbox(label="Summary", lines=10)
        btn1.click(summarize_document, pdf, out1)
    with gr.Tab("Ask a Question"):
        q = gr.Textbox(label="Your question")
        btn2 = gr.Button("Get Answer")
        out2 = gr.Textbox(label="Answer")
        btn2.click(ask_document, [pdf, q], out2)

app.launch(share=True, debug=True)
```
